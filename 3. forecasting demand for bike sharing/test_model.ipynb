Этот ноутбук нужен для быстрой проверки, что модель `model.joblib` корректно загружается в отдельной среде вместе с кастомными классами из файла `features.py`.

```python
from pathlib import Path

import joblib
import pandas as pd
from sklearn.metrics import mean_squared_error

# Важно импортировать features, чтобы joblib смог восстановить кастомные классы
import features
```

```python
TARGET_COL = 'Rented Bike Count'

base_path = Path('.')
test_path = base_path / 'test.csv'
model_path = base_path / 'model.joblib'

df_test = pd.read_csv(test_path)
X_test = df_test.drop(columns=TARGET_COL)
y_test = df_test[TARGET_COL]
```

```python
# Импорт модели и метаданных из файла
artifact_loaded = joblib.load(model_path)

model_loaded = artifact_loaded['model']
metadata_loaded = artifact_loaded['metadata']

y_pred = model_loaded.predict(X_test)

print(
    'RMSE загруженной модели: ',
    mean_squared_error(y_test, y_pred) ** 0.5
)

# Пример доступа к метаданным
print('Метаданные модели', metadata_loaded)
print('Все ок: модель успешно загружена и сделала предсказания.')
```
